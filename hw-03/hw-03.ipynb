## Задание 1. Средние буквы слова

Дано слово из латинских букв. Выводим среднюю букву, если число букв нечётное, и две средних буквы, если чётное.

```python
word = 'test'

n = len(word)
middle = n // 2
if n % 2 == 0:
    print(word[middle - 1:middle + 1])
else:
    print(word[middle])
```

## Задание 2 (выполнять не обязательно). Подбор пар

Сортируем оба списка по алфавиту и знакомим людей с одинаковыми индексами. Если длина списков разная — знакомить никого не будем.

```python
boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']

boys_sorted = sorted(boys)
girls_sorted = sorted(girls)

if len(boys_sorted) != len(girls_sorted):
    print('Внимание, кто-то может остаться без пары!')
else:
    print('Идеальные пары:')
    for boy, girl in zip(boys_sorted, girls_sorted):
        print(boy, 'и', girl)
```
