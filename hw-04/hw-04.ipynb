Программа-помощник секретаря: по номеру документа выводит владельца (команда **p**) или полку хранения (команда **s**); завершение работы — команда **q**.

## Исходные данные

```python
documents = [
    {'type': 'passport', 'number': '2207 876234', 'name': 'Василий Гупкин'},
    {'type': 'invoice', 'number': '11-2', 'name': 'Геннадий Покемонов'},
    {'type': 'insurance', 'number': '10006', 'name': 'Аристарх Павлов'},
]

directories = {
    '1': ['2207 876234', '11-2'],
    '2': ['10006'],
    '3': [],
}
```

## Функции

- `find_document(number)` — ищет документ по номеру;
- `find_shelf(number)` — ищет полку по номеру документа;
- `show_owner()` — обработка команды «p»;
- `show_shelf()` — обработка команды «s»;
- `main()` — цикл обработки команд до ввода «q».

Глобальные данные — только `documents` и `directories` (по условию).

```python
def find_document(number):
    """Возвращает документ с заданным номером или None."""
    for document in documents:
        if document['number'] == number:
            return document
    return None


def find_shelf(number):
    """Возвращает полку, на которой хранится документ, или None."""
    for shelf, numbers in directories.items():
        if number in numbers:
            return shelf
    return None


def show_owner():
    """Команда «p»: выводит владельца документа по его номеру."""
    number = input('Введите номер документа:
')
    document = find_document(number)
    if document is None:
        print('Владелец документа: владелец не найден')
    else:
        print('Владелец документа:', document['name'])


def show_shelf():
    """Команда «s»: выводит полку, на которой хранится документ."""
    number = input('Введите номер документа:
')
    shelf = find_shelf(number)
    if shelf is None:
        print('Документ не найден')
    else:
        print('Документ хранится на полке:', shelf)


def main():
    """Обрабатывает команды пользователя, пока не введена «q»."""
    while True:
        command = input('Введите команду:
')
        if command == 'p':
            show_owner()
        elif command == 's':
            show_shelf()
        elif command == 'q':
            break
```
