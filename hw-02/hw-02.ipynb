## Задание 1. Високосный год

Дана переменная с четырёхзначным числом (год). Программа определяет, является ли год високосным или обычным.

Год високосный, если он делится на 4, но не делится на 100, либо делится на 400.

```python
year = 2024

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print('Високосный год')
else:
    print('Обычный год')
```

```python
year = 2023

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print('Високосный год')
else:
    print('Обычный год')
```

## Задание 2 (выполнять не обязательно). Счастливый билет

Дана переменная с шестизначным числом (номер проездного билета). Билет счастливый, если сумма первых трёх цифр равна сумме последних трёх.

```python
number = 123456

digits = str(number)
if sum(map(int, digits[:3])) == sum(map(int, digits[3:])):
    print('Счастливый билет')
else:
    print('Несчастливый билет')
```
