## Подготовка: загрузка данных и функции построения графиков

```python
import matplotlib.pyplot as plt
import pandas as pd
import plotly.express as px
import seaborn as sns


def load_clients(path, limit=15):
    """Загружает первые limit строк с данными покупателей."""
    return pd.read_csv(path).head(limit)


def plot_scatter_seaborn(df):
    """Диаграмма разброса «возраст — расходы» (Seaborn)."""
    axes = sns.scatterplot(data=df, x='age', y='bill')
    axes.set_title('Возраст клиента и его расходы (Seaborn)')
    plt.show()


def plot_scatter_plotly(df):
    """Диаграмма разброса «возраст — расходы» (Plotly)."""
    figure = px.scatter(df, x='age', y='bill',
                        title='Возраст клиента и его расходы (Plotly)')
    figure.show(renderer='png')


def plot_regression(df):
    """Диаграмма рассеяния с линией регрессии (Seaborn regplot)."""
    axes = sns.regplot(data=df, x='age', y='bill')
    axes.set_title('Линия регрессии: возраст — расходы')
    plt.show()
```

## Задание 1. Диаграмма разброса «возраст — расходы»

Строим scatterplot по первым 15 строкам: возраст клиента (`age`) — расходы (`bill`).

### Seaborn

```python
clients = load_clients('web_clients_correct.csv')
plot_scatter_seaborn(clients)
```

### Plotly

```python
plot_scatter_plotly(clients)
```

## Задание 2 (выполнять не обязательно). Линия регрессии

Seaborn `regplot` — диаграмма рассеяния с линией регрессии.

```python
plot_regression(clients)
```
