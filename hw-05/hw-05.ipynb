Задача: для каждого `user_id` из `visit_log.csv` определить категорию покупки по `purchase_log.txt` и записать визиты с покупками в `funnel.csv`.

Исходные файлы `visit_log.csv` и `purchase_log.txt` берутся из материалов к ДЗ и лежат рядом с ноутбуком. Содержимое `visit_log.csv` не помещается в память, поэтому он обрабатывается только построчно.

## Решение

```python
import json

# purchase_log.txt — небольшой словарь "user_id -> категория"
purchases = {}
with open('purchase_log.txt', 'r', encoding='utf-8') as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        record = json.loads(line)
        if record['user_id'] == 'user_id':  # строка-заголовок
            continue
        purchases[record['user_id']] = record['category']

print('категорий в словаре:', len(purchases))

# visit_log.csv — большой, обрабатываем только построчно
rows = 0
with open('visit_log.csv', 'r', encoding='utf-8') as visit_file,      open('funnel.csv', 'w', encoding='utf-8') as funnel_file:
    header = visit_file.readline().strip()
    funnel_file.write(header + ',category
')
    for line in visit_file:
        line = line.strip()
        if not line:
            continue
        user_id, source = line.split(',')
        if user_id in purchases:
            funnel_file.write(f'{user_id},{source},{purchases[user_id]}
')
            rows += 1

print('строк с покупкой в funnel.csv:', rows)
```

## Проверка результата

```python
# первые строки funnel.csv
with open('funnel.csv', 'r', encoding='utf-8') as f:
    for i, line in enumerate(f):
        if i >= 3:
            break
        print(line.strip())
```

## Результат

В файле `funnel.csv` рядом с ноутбуком — 13 798 визитов с покупками (заголовок + строки `user_id,source,category`). Визиты без покупок в `funnel.csv` не попадают, исходный `visit_log.csv` не изменяется.
