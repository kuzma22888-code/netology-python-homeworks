## Задание 1. Категория возраста через `.apply`

Загружаем данные (до 15 строк) и функцией `.apply` заменяем столбец `age` категорией: «Старше 25» — если возраст больше 25, иначе «Младше или равен 25».

```python
import pandas as pd


def load_clients(path, limit=15):
    """Загружает первые limit строк с данными покупателей."""
    return pd.read_csv(path).head(limit)


def age_category(age):
    """Возвращает текстовую категорию по возрасту."""
    if age > 25:
        return 'Старше 25'
    return 'Младше или равен 25'


def show_age_categories(df):
    """Заменяет столбец age категориями и выводит его."""
    df['age'] = df['age'].apply(age_category)
    print(df['age'].rename(None))


clients = load_clients('web_clients_correct.csv')
show_age_categories(clients)
```

## Задание 2 (выполнять не обязательно). Мужчины старше 30

Отбираем строки DataFrame по двум условиям: пол — мужской (`sex == 'male'`), возраст — больше 30.

```python
def filter_adult_men(df):
    """Отбирает мужчин старше 30 лет."""
    return df[(df['sex'] == 'male') & (df['age'] > 30)]


clients = load_clients('web_clients_correct.csv')
adult_men = filter_adult_men(clients)
print(adult_men[['name', 'sex', 'age']].to_string(index=False))
```
