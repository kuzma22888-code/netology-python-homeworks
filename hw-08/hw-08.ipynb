## Обязательная часть

Работа ведётся с набором данных о состоянии здоровья лошадей, испытывающих кишечные колики.

**Выбранные столбцы (8 = 4 числовых + 4 категориальных):**

Числовые:

### Задание 1. Базовое изучение

Для числовых столбцов считаю среднее, медиану, моду, std, min/max; для категориальных — моду с частотой, число уникальных значений и пропусков.

```python
import pandas as pd


def load_data(url):
    """Загружает данные о лошадях, '?' считается пропуском."""
    column_names = [
        'surgery', 'age', 'hospital_number', 'rectal_temp', 'pulse',
        'respiratory_rate', 'temp_of_extremities', 'peripheral_pulse',
        'mucous_membranes', 'capillary_refill_time', 'pain', 'peristalsis',
        'abdominal_distension', 'nasogastric_tube', 'nasogastric_reflux',
        'nasogastric_reflux_ph', 'rectal_exam_feces', 'abdomen',
        'packed_cell_volume', 'total_protein', 'abdominocentesis_appearance',
        'abdomcentesis_total_protein', 'outcome', 'surgical_lesion',
        'lesion_type_1', 'lesion_type_2', 'lesion_type_3', 'cp_data',
    ]
    return pd.read_csv(url, header=None, names=column_names, na_values='?')


def select_columns(df, columns):
    """Отбирает заданные столбцы."""
    return df[columns].copy()


def describe_numeric(series):
    """Базовые метрики числового столбца."""
    return {
        'среднее': round(series.mean(), 2),
        'медиана': series.median(),
        'мода': series.mode()[0],
        'std': round(series.std(), 2),
        'min': series.min(),
        'max': series.max(),
    }


def describe_categorical(series):
    """Базовые метрики категориального столбца."""
    counts = series.value_counts()
    return {
        'мода (самое частое значение)': counts.index[0],
        'частота моды': int(counts.iloc[0]),
        'уникальных значений': series.nunique(),
        'пропусков': int(series.isna().sum()),
    }


NUMERIC_COLUMNS = ['rectal_temp', 'pulse', 'respiratory_rate', 'packed_cell_volume']
CATEGORICAL_COLUMNS = ['surgery', 'age', 'temp_of_extremities', 'pain']
SELECTED = NUMERIC_COLUMNS + CATEGORICAL_COLUMNS

df = load_data('https://raw.githubusercontent.com/obulygin/pyda_homeworks/'
               'master/statistics_basics/horse_data.csv')
selected = select_columns(df, SELECTED)

print('=== Задание 1: базовые метрики ===')
print('\nРазмер выборки:', len(selected), 'строк\n')
for column in NUMERIC_COLUMNS:
    series = selected[column]
    print(f'[{column}] числовой')
    for name, value in describe_numeric(series).items():
        print(f'  {name}: {value}')
    print()

for column in CATEGORICAL_COLUMNS:
    series = selected[column]
    print(f'[{column}] категориальный')
    for name, value in describe_categorical(series).items():
        print(f'  {name}: {value}')
    print()
```

**Краткие результаты:**

- `rectal_temp`: среднее 38.17 ≈ медиана 38.2, мода 38 — около нормы 37.8; разброс небольшой (std 0.73), но min 35.4 и max 40.8 — клинические состояния (шок и инфекция).
- `pulse`: среднее 71.9 заметно выше нормы 30–40 (медиана 64, max 184) — у лошадей с коликами пульс учащён из-за боли и шока.
- `respiratory_rate`: среднее 30.4 против нормы 8–10 — та же картина; по описанию столбца его информативность сомнительна из-за сильных колебаний.
- `packed_cell_volume`: среднее 46.3 в верхней части нормы 30–50 — признак дегидратации части животных.
- `surgery`: мода 1 — у большинства (180 из 300) была операция.
- `age`: мода 1 — подавляющее большинство (276 из 300) взрослые лошади.
- `temp_of_extremities`: мода 3 («прохладные») — типичный признак нарушения периферического кровообращения при коликах.
- `pain`: мода 3 («периодическая умеренная боль») — самый частый уровень боли.

### Задание 2. Работа с выбросами

Выбросы ищу правилом межквартильного размаха (IQR): значения вне `[Q1 − 1.5·IQR; Q3 + 1.5·IQR]`.

```python
def count_outliers_iqr(series):
    """Число выбросов по правилу межквартильного размаха."""
    q1, q3 = series.quantile([0.25, 0.75])
    iqr = q3 - q1
    low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    outliers = series[(series < low) | (series > high)]
    return low, high, len(outliers)


print('=== Задание 2: выбросы (IQR) ===')
for column in NUMERIC_COLUMNS:
    low, high, count = count_outliers_iqr(selected[column])
    print(f'{column}: границы [{low:.2f}; {high:.2f}], выбросов {count}')
```

**Гипотезы о причинах и решение:**

- `rectal_temp` (14 выбросов вне 36.75–39.55): низкие значения — поздний шок (по описанию температура падает в шоке), высокие — инфекционный процесс. Это **реальные клинические состояния, а не ошибки измерений**.
- `pulse` (5 выбросов выше 148): экстремальная тахикардия при тяжёлом шоке и сильной боли.
- `respiratory_rate` (17 выбросов выше 62.25): тяжёлый дистресс и боль; дополнительно сам столбец малонадёжен по описанию.
- `packed_cell_volume` (3 выброса вне 17–73): крайние степени дегидратации.

**Решение: выбросы сохраняем.** Все они физиологически объяснимы и клинически информативны — именно крайние значения отличают тяжёлых пациентов, их удаление исказило бы данные. Поэтому пропуски ниже заполняем медианой, а не средним — медиана устойчива к выбросам.

### Задание 3. Работа с пропусками

Считаю количество пропусков (`NaN` после замены `?`) в выбранных столбцах; заполняю их: числовые — медианой, категориальные — модой.

```python
def count_missing(df):
    """Число пропусков по каждому столбцу."""
    return df.isna().sum()


def fill_missing(df, numeric_columns):
    """Заполняет пропуски: числовые — медианой, категориальные — модой."""
    filled = df.copy()
    for column in df.columns:
        if column in numeric_columns:
            filled[column] = filled[column].fillna(filled[column].median())
        else:
            filled[column] = filled[column].fillna(filled[column].mode()[0])
    return filled


print('=== Задание 3: пропуски в выбранных столбцах ===')
missing = count_missing(selected)
print(missing.to_string())

filled = fill_missing(selected, NUMERIC_COLUMNS)
print('\nпропусков после заполнения:', int(filled.isna().sum().sum()))
```

**Результат и обоснование методов:** наибольшие пропуски — `rectal_temp` (60), `respiratory_rate` (58), `temp_of_extremities` (56), `pain` (55): субъективные и аппаратные измерения выполнялись не на всех этапах наблюдения.

Числовым столбцам заполняю медианой — устойчива к выбросам, которые мы осознанно сохранили (среднее сдвинулось бы); категориальным — модой, так как «среднего» у категории нет, а мода — наиболее типичное состояние. Итоговый датафрейм пропусков не содержит: `isna().sum().sum() == 0`.
